```python
# Understanding python yield function
# The function expects a number and counts the number from 0 to that number. The result is shown in a list.
def counting_non_yield_example(num):
    i = 0
    numbers = []
    while i < num:
        numbers.append(i+1)
        i += 1
    return numbers
```

```python
print(counting_non_yield_example(10))
```

Notice in the above example the program is executed and it returns all the number at once. If anytime we need to return the count to a very large number, that since the program executes everything at once, it will run for a very long time and will show the results together. This will be due to nature of return keyword and the high memory requirement of the program.

```python
#yield example of the counting function
from typing import Iterator

def counting_with_yield(n: int)-> Iterator[int]:
    i = 1
    while i < n:
        yield i
        i += 1
```

```python
count = counting_with_yield(10)
```
